```python
#FEATURE ENGINEERING

from sklearn.model_selection import train_test_split   #to split data into train and test 
from sklearn.preprocessing import StandardScaler       #for scvalinng the data
from sklearn.linear_model import LogisticRegression    #for modeling the data 
from sklearn.metrics import classification_report, confusion_matrix     #for model evaluation   
from sklearn.pipeline import Pipeline       #for training and doing the tranformation operations on model
from sklearn.compose import ColumnTransformer       
from sklearn.preprocessing import OneHotEncoder
```

```python
df.head()
```

```python
df_model = df.drop(["nameOrig", "nameDest", "isFlaggedFraud"], axis = 1)
```

```python
df_model.head()
```

```python
categorical = ["type"]
numeric = ["amount", "oldbalanceOrg", "newbalanceOrig", "oldbalanceDest", "newbalanceDest"]
```

```python
X = df_model.drop("isFraud", axis = 1)  #axis = 1---> column
y = df_model["isFraud"]
```

```python
X_train, X_test, y_train, y_test =  train_test_split(X,y, test_size=0.3, stratify=y)
```

```python
preprocessor = ColumnTransformer(
    transformers = [
        ("num", StandardScaler(), numeric),
        ("cat", OneHotEncoder(drop="first"), categorical)
    ],
    remainder="drop"

    
)
```

```python
pipeline = Pipeline([
    ("prep", preprocessor),
    ("clf", LogisticRegression(class_weight="balanced", max_iter = 1000))

])
```

```python
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)
```

```python
classification_report(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```

```python
confusion_matrix(y_test,y_pred)
```

```python
pipeline.score(X_test,y_test) *100
```

```python
#Exporting the pipeline
import joblib

joblib.dump(pipeline,"fraud_detection_pipeline.pkl")

#we will used this in the relevant website for the detection
```
